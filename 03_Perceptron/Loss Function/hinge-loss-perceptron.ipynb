```python
from sklearn.datasets import make_classification
import numpy as np
X, y = make_classification(n_samples=100, n_features=2, n_informative=1,n_redundant=0,
                           n_classes=2, n_clusters_per_class=1, random_state=41,hypercube=False,class_sep=15)
```

```python
y.shape
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(10,6))
plt.scatter(X[:,0],X[:,1],c=y,cmap='winter',s=100)
```

```python
def perceptron(X,y):
    
    w1=w2=b=1
    lr = 0.1
    
    for j in range(1000):                            # 1000 iterations is the epochs
    
        for i in range(X.shape[0]):                  # loop through all the samples

            # check condition
            z = w1*X[i][0] + w2*X[i][1] + b          # compute the linear combination Formula: z = w1*x1 + w2*x2 + b

            if z*y[i] < 0:                           # if the condition is not satisfied, update the weights and bias
                w1 = w1 + lr*y[i]*X[i][0]            # update weight w1 , Formula: w1 = w1 + lr*y[i]*x1
                w2 = w2 + lr*y[i]*X[i][1]            # update weight w2 , Formula: w2 = w2 + lr*y[i]*x2
                b = b + lr*y[i]                      # update bias b , Formula: b = b + lr*y[i]
            
    return w1,w2,b                                   # return the weights and bias after training
            
```

```python
w1,w2,b = perceptron(X,y)
```

```python
b
```

```python
m = -(w1/w2)
c = -(b/w2)

print(m,c)
```

```python
x_input = np.linspace(-3,3,100)
y_input = m*x_input + c

plt.figure(figsize=(10,6))
plt.plot(x_input,y_input,color='red',linewidth=3)
plt.scatter(X[:,0],X[:,1],c=y,cmap='winter',s=100)
plt.ylim(-3,2)
```
